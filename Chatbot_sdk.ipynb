```python
import os
from dotenv import load_dotenv
from pypdf import PdfReader
from agents import Agent, Runner, set_default_openai_api, set_tracing_disabled
```

```python
load_dotenv(override=True)
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("API_TOKEN")
os.environ["OPENAI_BASE_URL"] = "https://openrouter.ai/api/v1"
set_default_openai_api("chat_completions")
set_tracing_disabled(True)
```

```python
reader = PdfReader("Resources/Jude Enang.pdf")
prof_summary = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        prof_summary += text + "\n"

print(len(prof_summary))
```

```python
name = "Jude Enang"

instructions = f"""You are acting as {name}, representing {name} on their website.
Answer questions about {name}'s career, background, skills, and experience, in the first person.
Be professional, engaging, and approachable, as if speaking to a potential client or employer.
Be specific: use real companies, job titles, dates, and achievements from the CV below.
If the CV doesn't cover something, say so honestly instead of guessing.

## CV:
{prof_summary}
"""

cv_agent = Agent(
    name="CV Agent",
    instructions=instructions,
    model="openai/gpt-4o-mini",
)
```

```python
async def live_chat(message, history):
    conversation = [{"role": h["role"], "content": h["content"]}
                    for h in history]
    conversation.append({"role": "user", "content": message})

    result = await Runner.run(cv_agent, conversation)
    return result.final_output

gr.ChatInterface(
    live_chat,
    title="Chat with Jude Enang",
    description="Ask me anything about my career, background, skills, and experience!",
).launch()
```
